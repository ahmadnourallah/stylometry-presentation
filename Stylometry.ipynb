First we need to import the stylometry library:

```python
from faststylometry import Corpus
from faststylometry import load_corpus_from_folder
from faststylometry import tokenise_remove_pronouns_en
from faststylometry import calculate_burrows_delta
from faststylometry import predict_proba, calibrate
```

The library depends on NLTK (the Natural Language Toolkit), so the first time that you are using it, you may need to run the following commands in your Python environment if you want to use the inbuilt tokeniser:

```python
import nltk
nltk.download("punkt")
```

We need to add the training data:

```python
train_corpus = Corpus()
train_corpus.add_book("JK Rowling", "Harry Potter 1 - Sorcerer's Stone", open("./texts/HarryPotterSorcerersStone.txt").read())
train_corpus.add_book("JK Rowling", "Harry Potter 2 - The Chamber Of Secrets", open("./texts/HarryPotterTheChamberOfSecrets.txt").read())
train_corpus.add_book("JK Rowling", "Harry Potter 3 - Prisoner of Azkaban", open("./texts/HarryPotterPrisonerofAzkaban.txt").read())
train_corpus.add_book("Tolkien", "The Hobbit", open("./texts/TheHobbit.txt").read())
train_corpus.add_book("Tolkien", "Lord of the Rings", open("./texts/LordOfTheRings.txt").read())
train_corpus.tokenise(tokenise_remove_pronouns_en)
```

Now we add the unknown book:

```python
test_corpus = Corpus()
test_corpus.add_book("", "The Cuckoo's Calling", open("./texts/TheCuckoosCalling.txt").read())
test_corpus.tokenise(tokenise_remove_pronouns_en)
```

```python
calculate_burrows_delta(train_corpus, test_corpus, vocab_size = 50)
```
